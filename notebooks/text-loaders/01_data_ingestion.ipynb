# Stage 1 — Data Ingestion

## 1.1 Understanding Data Ingestion and LangChain Document

##### below code snippet describes about how can we create a simple LangChain document and use it for RAG

````python
from langchain_core.documents import Document

# Create a Document object with page content and metadata. This is a single document, but you can create
# multiple documents and store them in a list or other data structure if needed.
document = Document(
    page_content="Azure Event Hubs is a real-time data streaming platform.",
    metadata={
        "source": "azure_event_hub.txt",
        "topic": "Azure Event Hubs",
        "category": "Azure"
    }
)
print(document)
# print("Type:", type(document))
# print("Content:", document.page_content)
# print("Metadata:", document.metadata)
````

````python
# multiple documents creation
documents = [
    Document(
        page_content="Azure Event Hubs is a real-time data streaming platform.",
        metadata={"source": "event_hub.pdf", "page": 1}
    ),
    Document(
        page_content="Azure Data Explorer is a fast and scalable data analytics service.",
        metadata={"source": "data_explorer.pdf", "page": 1}
    ),
]
````

````python
print('Number Of Documents:', len(documents))
print('Type : ',type(documents))
````

##### print all the LangChain Documents created in above cell

````python
# View all documents
for i, doc in enumerate(documents):
    print(f"Document {i + 1}")
    print("Content:", doc.page_content)
    print("Metadata:", doc.metadata)
    print("-" * 60)
````

# 2. What is Data Ingestion?

In simple terms:

> **Data ingestion is the process of taking information from an external source and bringing it into our RAG pipeline in a form that the pipeline can work with.**

Suppose your company has:

```text
PDF
DOCX
HTML
TXT
CSV
JSON
Web pages
Database
Azure Storage
```

These are **sources of knowledge**.

Our RAG application needs to transform these sources into something it understands.

Conceptually:

```text
                    External World
                         │
       ┌─────────────────┼─────────────────┐
       ▼                 ▼                 ▼
      PDF               DOCX             HTML
       │                 │                 │
       ▼                 ▼                 ▼
    PDF Loader        DOCX Loader       HTML Loader
       │                 │                 │
       └─────────────────┼─────────────────┘
                         ▼
                LangChain Documents
                         │
                         ▼
                  Next RAG Stage
```

This is where LangChain's `Document` becomes important.

# 3. What is a LangChain `Document`?

A LangChain `Document` is essentially a container for:

```text
┌───────────────────────────────────┐
│          Document                 │
│                                   │
│  page_content                     │
│  ─────────────────────────────    │
│  Actual textual content           │
│                                   │
│  metadata                         │
│  ─────────────────────────────    │
│  Information about the content    │
│                                   │
│  id                               │
│  ─────────────────────────────    │
│  Optional document identifier     │
└───────────────────────────────────┘
```

The two most important concepts initially are:

### `page_content`

The actual text.

### `metadata`

Information **about** that text.

---

The PDF itself is **not yet the thing we retrieve**.

The extracted textual representation becomes part of our RAG data pipeline.

# 9. `Document` is NOT the same as a physical document

This distinction is critical.

When we say:

```text
Document
```

in LangChain, don't automatically think:

```text
PDF file
DOCX file
```

A LangChain `Document` is a **data structure representing content**.

For example, one PDF can produce multiple LangChain Documents.

Imagine:

```text
azure.pdf
│
├── Page 1
├── Page 2
├── Page 3
└── Page 4
```

A loader might produce:

```text
List[Document]
│
├── Document(page 1)
├── Document(page 2)
├── Document(page 3)
└── Document(page 4)
```

Therefore:

> **One physical file does not necessarily equal one LangChain `Document`.**

This becomes particularly important when we learn chunking.

---

# 10. Document vs Chunk

This is another distinction we need to establish now.

Imagine:

```text
PDF
 │
 ▼
Loader
 │
 ▼
Documents
 │
 ▼
Chunking
 │
 ▼
Chunks
```

For example:

```text
azure_event_hub.pdf
        │
        ▼
    PDF Loader
        │
        ▼
  10 Documents
        │
        ▼
     Chunking
        │
        ▼
   80 Chunks
```

So:

```text
Physical file
      ↓
LangChain Document(s)
      ↓
Chunk(s)
      ↓
Embedding(s)
      ↓
Vector Index
```

We'll spend a **whole Stage 3** understanding chunking, so don't worry about the details yet.

# 11. Why do loaders return `Document` objects?

Now we can understand the purpose of LangChain's loader abstraction.

Suppose we have:

```text
PDF
DOCX
HTML
TXT
CSV
```

Without a common representation, every downstream component would need to understand every file format.

That would become messy:

```text
PDF → special processing
DOCX → special processing
HTML → special processing
CSV → special processing
...
```

Instead:

```text
PDF ───────┐
DOCX ──────┤
HTML ──────┤
TXT ───────┤
CSV ───────┘
     │
     ▼
Document
     │
     ▼
Common downstream pipeline
```

This is one of the major reasons abstractions like `Document` are useful.

# 16. One important terminology issue

You will encounter several terms:

```text
Source
Document
Page
Chunk
Embedding
Vector
```

They are **not interchangeable**.

For now, remember:

| Term      | Meaning                                          |
| --------- | ------------------------------------------------ |
| Source    | Original knowledge source                        |
| File      | Physical representation such as PDF/DOCX         |
| Document  | LangChain representation of content              |
| Page      | A physical page, often represented by a Document |
| Chunk     | Smaller piece created during chunking            |
| Embedding | Numerical representation of text                 |
| Vector    | The numerical embedding representation           |

The exact relationship depends on the loader and pipeline.
