## 1.2 Text File Ingestion

#### Objective

Load a real text knowledge-base file using a LangChain document loader and inspect the resulting Document object.

```python
from langchain_community.document_loaders import TextLoader
```

```python
# path to the text file
file_path = "../data/raw/txt/azure_event_hubs_knowledge.txt"
```

```python
# loader to read the text file and create a Document object
loader = TextLoader(file_path)
# load the document from the text file
documents = loader.load()
print( 'Number of documents loaded:', len(documents) )
```

```python
# Inspect the Document
print(documents[0])
```

```python
print(documents[0].page_content)
```

```python
print(documents[0].metadata)
```

```python
# Inspect the complete object
document = documents[0]

print("Document type:", type(document))
print()
print("Content type:", type(document.page_content))
print()
print("Metadata type:", type(document.metadata))
```

```python
documents[0].metadata.get("source")
```

```python
len(documents[0].page_content)
```

### Observations

1. The physical TXT file was loaded using LangChain's TextLoader.
2. The loader returned a list of LangChain Document objects.
3. The Document contains page_content and metadata.
4. The content contains the extracted text from the source file.
5. Metadata contains information about the source.
6. The downstream RAG pipeline can work with the Document abstraction instead of directly dealing with the TXT file.

###What I want you to understand before we move on

At this point you should be able to clearly answer:

Q1. What is Data Ingestion?

Bringing knowledge from an external source into the RAG pipeline.

Q2. What is a Loader?

A LangChain component that loads content from a particular type of source and converts it into Document objects.

Q3. What is a LangChain Document?

A common representation of content containing page_content and metadata.

Q4. What does TextLoader.load() return?

A list of LangChain Document objects.

Q5. Why is Document important?

It provides a common representation that downstream RAG components can work with regardless of the original source format.
