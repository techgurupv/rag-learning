```python
from langchain_community.document_loaders import (
    PyPDFLoader,
    PyMuPDFLoader,
)
from pathlib import Path
```

```python
pdf_path = Path("../../data/raw/pdf/azure_event_hubs_knowledge.pdf")

print("PDF:", pdf_path.resolve())
print("Exists:", pdf_path.exists())
```

### Create a PyPDFLoader

```python
pypdf_loader = PyPDFLoader(str(pdf_path))
pypdf_documents = pypdf_loader.load()
print("PyPDFLoader")
print("Number of documents:", len(pypdf_documents))
```

### Inspect PyPDFLoader metadata

```python
for document in pypdf_documents:
    print(document.metadata)
```

### Now inspect the first document:

```python
print(pypdf_documents[0].page_content[:500])
```

```python
print(pypdf_documents[0].metadata)
```

### Now use PyMuPDFLoader

```python
pymupdf_loader = PyMuPDFLoader(str(pdf_path))
pymupdf_documents = pymupdf_loader.load()
```

```python
print("PyMuPDFLoader")
print("Number of documents:", len(pymupdf_documents))
```

### Inspect PyMuPDF metadata

```python
for document in pymupdf_documents:
    print(document.metadata)
```

```python
print(pymupdf_documents[0].page_content[:500])
```

```python
print(pymupdf_documents[0].metadata)
```

### First comparison — document count

```python
print("PyPDFLoader:", len(pypdf_documents))
print("PyMuPDFLoader:", len(pymupdf_documents))
```

### Compare page content

```python
pypdf_text = pypdf_documents[0].page_content
pymupdf_text = pymupdf_documents[0].page_content
```

```python
print("PyPDFLoader:")
print(pypdf_text)
```

```python
print("PyMuPDFLoader:")
print(pymupdf_text)
```

### Compare character counts

```python
print(
    "PyPDFLoader characters:",
    len(pypdf_text)
)

print(
    "PyMuPDFLoader characters:",
    len(pymupdf_text)
)
```

### Compare metadata side by side

```python
print("PyPDFLoader metadata:")
print(pypdf_documents[0].metadata)

print()

print("PyMuPDFLoader metadata:")
print(pymupdf_documents[0].metadata)
```

### Create a comparison function

```python
def compare_pdf_loaders(name, documents):
    print("=" * 70)
    print(name)
    print("=" * 70)

    print("Number of documents:", len(documents))

    for index, document in enumerate(documents):
        print(
            f"Document {index + 1}: "
            f"{len(document.page_content)} characters"
        )

        print("Metadata:", document.metadata)

        print()
```

```python
compare_pdf_loaders(
    "PyPDFLoader",
    pypdf_documents
)
```

```python
compare_pdf_loaders(
    "PyMuPDFLoader",
    pymupdf_documents
)
```

### Compare all pages

```python
for index in range(
    min(len(pypdf_documents), len(pymupdf_documents))
):
    pypdf_length = len(
        pypdf_documents[index].page_content
    )

    pymupdf_length = len(
        pymupdf_documents[index].page_content
    )

    print(
        f"Page {index + 1}: "
        f"PyPDF={pypdf_length}, "
        f"PyMuPDF={pymupdf_length}"
    )
```

## PDF Loader Comparison

| Feature | PyPDFLoader | PyMuPDFLoader |
|---|---|---|
| LangChain integration | Yes | Yes |
| Underlying library | pypdf | PyMuPDF |
| Page-based Documents | Yes | Yes |
| Text extraction | Yes | Yes |
| Page metadata | Yes | Yes |
| Simple text PDFs | Good | Good |
| Complex PDF handling | Depends on document | Often more capable |
| Performance | Good | Generally very fast |
| OCR for scanned PDFs | Not inherently provided | Not inherently provided |
| Best choice | Depends on workload | Depends on workload |

### important conclusion

Don't memorize:

"PyMuPDF is better."

Instead remember:

PDF loader selection is an engineering decision based on the characteristics of your documents and the requirements of your RAG system.

The decision should consider:

Extraction quality
       +
Performance
       +
Metadata
       +
Tables
       +
Images
       +
OCR requirements
       +
Document layout
       +
Scale
       ↓
Loader selection
