### This Notebook extracts the images of a PDF file

```python
from pathlib import Path
import pymupdf


def extract_raw_images(pdf_path: str, output_dir: str = "extracted_images") -> None:
    out_dir = Path(output_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    extracted_count = 0

    # Using context manager for automatic resource cleanup
    with pymupdf.open(pdf_path) as doc:
        for page_idx, page in enumerate(doc):
            images = page.get_images(full=True)

            for img_idx, img in enumerate(images):
                xref = img[0]
                base_image = doc.extract_image(xref)

                image_bytes = base_image["image"]
                image_ext = base_image["ext"]  # e.g., 'png', 'jpeg'

                output_file = (
                    out_dir
                    / f"page_{page_idx + 1}_img_{img_idx + 1}.{image_ext}"
                )
                output_file.write_bytes(image_bytes)

                extracted_count += 1
                print(f"Extracted: {output_file.name}")

    print(f"\nDone: {extracted_count} image(s) saved to '{output_dir}'.")


```

```python
# Run the function
extract_raw_images("D:/AI Learning/rag-learning/data/raw/pdf/azure_event_hubs_scanned.pdf")
```
