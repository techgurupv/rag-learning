```python
from pathlib import Path
from docling.document_converter import DocumentConverter
```

```python
pdf_path = Path("D:/AI Learning/rag-learning/data/raw/pdf/azure_event_hubs_scanned.pdf")

print("Exists:", pdf_path.exists())
print("Path:", pdf_path)
```

```python
converter = DocumentConverter()
```

```python
result = converter.convert(pdf_path)

print("Status:", result.status)
```

```python
doc = result.document

print(type(doc))
```

```python
markdown_text = doc.export_to_markdown()

print(markdown_text)
```

```python
doc_dict = doc.export_to_dict()
print(type(doc_dict))
```

```python
import json

json_text = json.dumps(
    doc_dict,
    indent=2,
    ensure_ascii=False
)
print(json_text[:3000])
```

```python
from pathlib import Path

json_path = Path("output/scanned_document_docling.json")

# Create the parent directory (and any missing intermediate folders) if it doesn't exist
json_path.parent.mkdir(parents=True, exist_ok=True)

json_path.write_text(json_text,encoding="utf-8")
print(f"JSON saved to: {json_path}")
```

So our corrected pipeline is:
